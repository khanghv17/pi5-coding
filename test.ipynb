```python
import cv2
import numpy as np

# Tạo 60 frames giả
frames = [np.random.randint(0, 255, (240, 320, 3), dtype=np.uint8) for _ in range(60)]

fourcc = cv2.VideoWriter_fourcc(*'XVID')
output = cv2.VideoWriter('test_video.avi', fourcc, 15, (320, 240))

print(f"isOpened: {output.isOpened()}")

for frame in frames:
    output.write(frame)
    
output.release()

import os
print(f"File exists: {os.path.exists('test_video.avi')}")
print(f"File size: {os.path.getsize('test_video.avi')} bytes")
```

```python
import onnxruntime as ort
import numpy as np

# Load model
session = ort.InferenceSession("ctrgcn.onnx")

# Chuẩn bị input - shape (1, 3, 60, 33)
# 1 = batch, 3 = channels (x,y,z), 60 = frames, 33 = joints
input_data = np.random.randn(1, 3, 60, 33).astype(np.float32)

# Inference
outputs = session.run(None, {'input': input_data})
predictions = outputs[0]

# Lấy class có xác suất cao nhất
predicted_class = np.argmax(predictions, axis=1)
print(f"Predicted class: {predicted_class}")
```
